# **Series from Lists**


Common Uses of a Series (Column-Level Operations)

A Series is used when you isolate a single variable or perform mathematical operations across an array of data.

• Time-Series Analysis: Tracking a single metric over time, such as daily stock prices, sensor temperature readings, or website traffic.

• Vectorized Math Operations: Performing instant mathematical calculations across an entire column without using slow Python for loops (e.g., df['Price'] * 1.06 to apply a 6% tax to every row).

• String Manipulation: Applying text clean-up operations to a text column, such as converting all email addresses to lowercase or extracting zip codes.

• Feature Extraction: Generating descriptive statistics for a single metric, like finding the .mean(), .median(), .max(), or .value_counts() of a specific column.

```python
#clip
'''
Clip functions is used to trim or remove values in the series or
data frame that are outside the specified lower or upper boundaries.

 Any value below the lower threshold is replaced by the
 lower boundary, and any value above the upper threshold is
  replaced by the upper boundary.
unlike between it keeps the size of the original array.

'''
print(subs.clip(50,200))

```

```python
#drop_duplicates

d1 = pd.Series([1,1,1,2,2,2,3,3,4,])
d1
```

```python
#drop duplicates with keeping last or first
'''delete the first or last duplicate value in the duplicate values.
like in in 3 1's[1,1,1] we can delete the 1 at index 0 or
the one at index 2.
By Default it deletes the first duplicate value.'''
print(d1)
print()
print(d1.drop_duplicates())
print()
print(d1.drop_duplicates(keep='first'))
print()
d1.drop_duplicates(keep='last')
```
