```python
#7. np.argmax()
'''
  Returns the index/indices of the max element
  in a array /axis of a array.
  Mostly used on 1D arrays.
'''

#on 1D
print(np.argmax(a))
print(a)
print()

#on 2D
print(np.argmax(b,axis=0))
print(b)
print()

#on 3D
print(np.argmax(b1))
print(b1)

```

```python
#8. np.argmin()
'''just like argmax, it finds the lowest element index/indices.
'''
#np.argmax()
'''
  Returns the index/indices of the max element
  in a array /axis of a array.
  Mostly used on 1D arrays.
'''

#on 1D
print(np.argmin(a))
print(a)

#on 2D
print(np.argmin(b,axis=0))
print(b)
print()

#on 3D
print(np.argmin(b1))
print(b1)


```

```python
#np.in1d

'''
tests whether each element of a 1-D array (ar1)
 is also present in a second array (ar2),
returning a boolean array of the same length.
'''
import numpy as np

test = np.array([0, 1, 2, 5, 0])
states = [0, 2]

# Test presence
mask = np.in1d(test, states)
print(mask)
# Output: array([ True, False,  True, False,  True])

# Extract matching elements
print(test[mask])
# Output: [0 2 0]

```
