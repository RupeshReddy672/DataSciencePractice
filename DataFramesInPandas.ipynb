# Filtering a DataFrame

```python
# final all the final winners

mask = ipl['MatchNumber']=='Final'
ipl[mask][['Season','WinningTeam']]

#or
#ipl[ipl['MatchNumber']=='Final'][['Season','WinningTeam']]
```

```python
# how many matches has csk won in kolkata
#ipl['WinningTeam']=='Chennai Super Kings'

'''First searching for the matches won by csk
 then after that filtering the matches played in kolkata.
'''
csk_matches = ipl[ipl['WinningTeam']=='Chennai Super Kings']
print((csk_matches[csk_matches['City']=='Kolkata']).shape[0])

'''or searcing in the reverse order :
 Matched  played in kolkata and won by csk
'''

#or


ipl[(ipl['WinningTeam']=='Chennai Super Kings') & (ipl['City']=='Kolkata')].shape[0]
```

```python
# how many super over finishes have occured

print(ipl.value_counts('SuperOver')['Y'])
print()
#or  one more preferable method.
print(ipl[ipl['SuperOver']=='Y'].shape[0])
print()
print(ipl[ipl['SuperOver']=='Y'].value_counts('SuperOver'))
print()

print(ipl['SuperOver'].value_counts()['Y'])
# feels more pythonic than the 1st one
''' method with the shape felt like easiest of all'''
```
