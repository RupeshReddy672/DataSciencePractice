###`Q-1:`
i. Create a DataFrame birds from the above dictionary data which has the index labels.

ii. Display basic information about the dataFrame.

iii. Show Alternate rows of the dataframe.

````python
# code here
import numpy as np
import pandas as pd
data = {'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills', 'Cranes'],
        'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4, 3.5],
        'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2, 2],
        'priority': ['yes', 'yes', 'no', np.nan, 'no', 'no', 'no', 'yes', 'no', 'no','yes']}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k']
````

````python
#1.1
birds = pd.DataFrame(data,index=labels)
birds
````

````python
#1.2Display basic information about the dataFrame.
birds.info()

````

````python
#1.3 Show Alternate rows of the dataframe.
birds[::2]
````

###`Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
# code here

def team1_vs_team2(team1 = 'Kolkata Knight Riders',team2 = 'Chennai Super Kings'):
  df = ipl[(((ipl['Team1']==team1)&(ipl['Team2']==team2)) | ((ipl['Team1']==team2)&(ipl['Team2']==team1)) )]


team1 = 'Kolkata Knight Riders'
team2 ='Chennai Super Kings'

df = ipl[(((ipl['Team1']==team1)&(ipl['Team2']==team2)) | ((ipl['Team1']==team2)&(ipl['Team2']==team1)) )]
print(f'Total Matches played: {df.shape[0]}')
print(f'Matches Won by KKR: {df[df['WinningTeam']==team1].shape[0]}')
print(f'Matches Won by CSK: {df[df['WinningTeam']==team2].shape[0]}')
print(f'player of the match in all matches award: {df['Player_of_Match'].value_counts().index[0]}')



````
